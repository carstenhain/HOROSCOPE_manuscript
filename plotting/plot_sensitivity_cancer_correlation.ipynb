```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("plotting_data/correlation_metrics.tsv", sep="\t")
df.head()
```

```python
ax1 = plt.subplot(1, 3, 1)
ax2 = plt.subplot(1, 3, 2)
ax3 = plt.subplot(1, 3, 3)


ax1.plot(df['c'], df['p_p'], marker='o', label="p-arm loss")
ax1.plot(df['c'], df['p_q'], marker='o', label="q-arm loss")
ax1.set_title("P-values")
ax1.legend()
ax1.set_ylim(0, 0.05)

ax2.plot(df['c'], df['r2_p'], marker='o', label="p-arm loss")
ax2.plot(df['c'], df['r2_q'], marker='o', label="q-arm loss")
ax2.legend()
ax2.set_title("R2 values")
ax2.set_ylim(0, 1)

ax3.plot(df['c'], df['n'], marker='o')
ax3.set_title("Number of arm losses")
ax3.set_ylim(440, 600)

for ax in [ax1, ax2, ax3]:
    ax.set_xlabel("Filtering threshold")

ax1.plot([0.3, 0.3], [0, 0.05], linestyle='--', color='gray')
ax2.plot([0.3, 0.3], [0, 1], linestyle='--', color='gray')
ax3.plot([0.3, 0.3], [440, 600], linestyle='--', color='gray')

plt.gcf().set_size_inches((10, 3))
plt.tight_layout()
plt.savefig("plots/sensitivity_cancer_correlation.png")
```
