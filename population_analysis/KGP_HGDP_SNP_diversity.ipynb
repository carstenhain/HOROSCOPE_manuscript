# Population SNP heterozygosity

For each population and variant site, calculate the biallelic SNP heterozygosity from the alternate-allele frequency $p = AC / AN$: $E = 2 * p * (1-p)$. The code below reports the mean SNP heterozygosity per population.

```python
import pandas as pd
import os
from tqdm import tqdm
```

```python
def calc_het_metrics (path_to_tsv:str):
    df = pd.read_csv(path_to_tsv, sep="\t")
    df = df[df["AN"] > 0]
    df["AF"] = df["AC"] / df["AN"]
    df["RF"] = 1 - df["AF"]
    df["HET"] = 2 * df["AF"] * df["RF"]

    df.groupby("POPULATION").agg({
        "HET": "mean"
    }).to_csv(path_to_tsv + "_het.tsv", sep="\t")
```

```python
chromosomes = [f"chr{i}" for i in range(1, 23)]
```

```python
for chrom in tqdm(chromosomes):
    path_to_data = f"/scratch/hain/pop_freq/{chrom}_population_ac_an.tsv"
    if not os.path.exists(path_to_data):
        continue
    df = pd.read_csv(path_to_data, sep="\t")
    df = df[df["AN"] > 0]
    df["AF"] = df["AC"] / df["AN"]
    df["RF"] = 1 - df["AF"]
    df["HET"] = 2 * df["AF"] * df["RF"]

    df.groupby("POPULATION").agg({
        "HET": "mean"
    }).to_csv(f"/scratch/hain/pop_freq/{chrom}_het.tsv", sep="\t", index=True)
```
